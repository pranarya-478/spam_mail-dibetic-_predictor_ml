```python
#loading  the data from csv  file to a pandas Dataframe
raw_mail_data = pd.read_csv('/mail_data.csv')
```

```python
print(raw_mail_data)
```

```python
#replace the null values with a null string
mail_data = raw_mail_data.where ((pd.notnull(raw_mail_data)),'')
```

```python
#printig the forst 5 rows of the Dataframe
mail_data.head()
```

```python
#checking the number of rows and columns in the dataframes
mail_data.shape
```

```python
# separating the data ss texts and label

X = mail_data['Message']
Y = mail_data['Category']
```

```python

# label spam mail as 0; ham mail as 1 ;

mail_data.loc[mail_data['Category'] == 'spam', 'Category',] = 0
mail_data.loc[mail_data['Category'] == 'ham', 'Category',] = 1
```

```python
print(X)
```

```python
print(Y)
```

```python
#splitting the data into training data & test data

X_train , X_test ,Y_train ,Y_test = train_test_split( X,Y,  test_size =0.2 , random_state = 3)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

```python
# transfrom the text data to feature vectors that can be used as input to the Logistic Regression

feature_extraction = TfidfVectorizer(min_df = 1, stop_words = 'english', lowercase = True)

X_train_features = feature_extraction.fit_transform(X_train)
X_test_features =  feature_extraction.transform(X_test)

#cover Y_train and Y_test values as integers
# Note: Make sure to clean up the 'ham' strings first if you get a ValueError during casting.
Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

Training the Model

Logictics Regrssion

```python
model = LogisticRegression()
```

```python
# traning the Logistic Regrssion model with the training data

model.fit(X_train_features , Y_train)
```

Evaluating the Trained Model

```python
# predicton on trainig data

prediction_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(Y_train , prediction_on_training_data)
```

```python
print('Accuracy on training data : ', accuracy_on_training_data)
```
